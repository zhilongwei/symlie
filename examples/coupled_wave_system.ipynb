# Lie Symmetry Analysis of Coupled Multi-Component Systems

This tutorial demonstrates using **`symlie`** to analyze multi-dependent field systems, taking the $2 \times 2$ first-order hyperbolic wave system as a canonical benchmark:
$$u_t - v_x = 0, \qquad v_t - u_x = 0$$

We investigate:
1. Multi-variable Fréchet operator matrices ($2 \times 2$).
2. The 12-dimensional symmetry space found inside a total-degree-1 polynomial ansatz for $(u, v)$.
3. Verification of coupled vector fields and cross-field rotation symmetries.

```python
import sympy as sp

from symlie import (
    frechet_derivative,
    infinitesimals,
    max_derivative_order,
    verify_generator,
)

sp.init_printing()

x, t = sp.symbols("x t")
u = sp.Function("u")(x, t)
v = sp.Function("v")(x, t)

# 2x2 first-order hyperbolic system
eq1 = u.diff(t) - v.diff(x)
eq2 = v.diff(t) - u.diff(x)

print("System Order:", max_derivative_order([eq1, eq2], [u, v], [x, t]))
display(sp.Eq(eq1, 0), sp.Eq(eq2, 0))
```

## 1. Multi-Component Fréchet Derivative

For multi-dependent fields, `frechet_derivative` produces the $2 \times 2$ matrix operator acting on test functions $(Q_u, Q_v)$:

```python
Q_u = sp.Function("Q_u")(x, t)
Q_v = sp.Function("Q_v")(x, t)

D_sys = frechet_derivative([eq1, eq2], [u, v], [x, t], [Q_u, Q_v])
print("Fréchet Linearization Matrix:")
display(D_sys)
```

## 2. Degree-1 Polynomial Symmetry Space

Solving with `infinitesimals(ansatz_degree=1)` derives a 12-dimensional polynomial-ansatz solution space. It is not the complete Lie point-symmetry algebra: because the system is linear, every solution pair $(h_1,h_2)$ supplies the superposition symmetry $h_1\partial_u+h_2\partial_v$.

```python
sol = infinitesimals([eq1, eq2], [u, v], [x, t], ansatz_degree=1)
print(f"Dimension within degree-1 ansatz: {sol.ansatz_dimension}\n")

for i, gen in enumerate(sol.basis, 1):
    is_valid = verify_generator([eq1, eq2], [u, v], [x, t], gen)
    print(f"X_{i:2d}: xi = {gen.xi},  phi = {gen.phi}  |  Verified: {is_valid}")

print("\nGeneral Infinitesimal Solution:")
display(sol.general)
```
